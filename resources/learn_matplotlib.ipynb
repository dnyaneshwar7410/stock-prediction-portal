## Subplots and Legends

```python
days = [1,2,3,4,5]
original_price = [10,20,25,33,77]
predicted_price = [8,20, 24, 34,76]
```

```python
# Legends
plt.plot(days, original_price,'b-o', label="Original Price")
plt.plot(days, predicted_price,'r-o',  label="Predicted Price")
plt.legend()
plt.title("Plot with Legends")
plt.xlabel("Weekdays")
plt.ylabel('Price')
```

## Subplots

```python
# First subplots - original price
plt.subplot(3,1,1)
plt.plot(days, original_price,'b-o', label="Original Price")
plt.title("Original prices")
plt.xlabel("Weekdays")
plt.ylabel('Price')
plt.legend()


# Second subplot predicted price
plt.subplot(3,1,2)
plt.plot(days, predicted_price,'r-o',  label="Predicted Price")
plt.legend()
plt.title("Predicted price over time")
plt.xlabel("Weekdays")
plt.ylabel('Price')



# Third subplot to calculate difference
difference = [ o-p for o,p in zip(original_price, predicted_price)]
plt.subplot(3, 1, 3)
plt.bar(days, difference, color='purple', label='Price Difference')
plt.title("Prce Difference between original and predicted")
plt.xlabel('Weekdays')
plt.ylabel('Price difference')
plt.legend();

plt.tight_layout()
```

## Saving the figure

```python
# Legends
plt.plot(days, original_price,'b-o', label="Original Price")
plt.plot(days, predicted_price,'r-o',  label="Predicted Price")
plt.legend()
plt.title("Plot with Legends")
plt.xlabel("Weekdays")
plt.ylabel('Price')
plt.savefig('example.png', dpi=300)  
```
